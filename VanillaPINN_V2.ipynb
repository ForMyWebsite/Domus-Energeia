```python

# =============================================================================
# CELL 3 -- GRAD06
# =============================================================================
path = find_grad_dataset("grad06")
result_grad06 = run_experiment(path, seed=SEED, n_epochs=N_EPOCHS, print_every=500)
results.append(result_grad06)
result_grad06

```
